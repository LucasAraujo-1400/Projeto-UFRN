```python
import pickle






#####################################################################
########## PROJETO - SISTEMA DE GESTÃO DE ESTACIONAMENTOS ###########
#####################################################################
def recuperar_clientes():
    try:
      clientes = {}
      arq_clientes = open("clientes.dat", "rb")
      clientes = pickle.load(arq_clientes)
      arq_clientes.close()
      # Normalizar os dados: garantir que todos os clientes tenham o status de ativo/inativo
      for nome, dados in clientes.items():
          if len(dados) == 3: # Se o status ativo está faltando
              clientes[nome].append(True) # Adiciona True por padrão
          elif len(dados) < 3: # Caso os dados estejam ainda mais incompletos
              while len(clientes[nome]) < 3:
                  clientes[nome].append('') # Preenche com vazios até 3
              clientes[nome].append(True) # Adiciona o status ativo
    except:
        clientes = {
          # Nome : Email, Data de nascimento, Celular
            "Lucas" : ["Lucas@ufrn.edu.br", "14/04/2006", "(84) 99999-2020", True],
            "Mário" : ["Mario@ufrn.edu.br", "23/01/2007", "(84) 99999-2121", True],
            "Pedro" : ["Pedro@ufrn.edu.br", "08/05/2008", "(84) 99999-2222", True]
        }
        arq_clientes = open("clientes.dat", "wb")
        pickle.dump(clientes, arq_clientes)
        arq_clientes.close()
    return clientes

def recuperar_locacoes():
    try:
         locacoes = {}
         arq_locacoes = open("locacoes.dat", "rb")
         locacoes = pickle.load(arq_locacoes)
         arq_locacoes.close()
    except:
        locacoes = {
           # CPF : Nome, E-mail, Data de nascimento, Celular
            "11111111111" : ["Lucas Araújo Diniz", "Lucas@ufrn.edu.br", "14/04/2006", "(84) 99999-2020", True],
            "22222222222" : ["Mário dos Santos", "Mario@ufrn.edu.br", "23/01/2007", "(84) 99999-2121", True],
            "33333333333" : ["Pedro Alves Cabral", "Pedro@ufrn.edu.br", "08/05/2008", "(94) 99999-2222", True]
        }
        arq_locacoes = open("locacoes.dat", "wb")
        pickle.dump(locacoes, arq_locacoes)
        arq_locacoes.close()
    return locacoes

def recuperar_vagas():
    try:
        vagas = {}
        arq_vagas = open("vagas.dat", "rb")
        vagas = pickle.load(arq_vagas)
    except:
        vagas = {
           # Placa : Carro, Cor, Nome, Celular
            "ABC1D23" : ["Porsche", "Branco", "Lucas Araújo Diniz", "(84) 99999-2020", True],
            "DEF4E56" : ["Ferrari", "Vermelho", "Mário dos Santos", "(84) 99999-2121", True],
            "GHI7F89" : ["Lamborghini", "Preto", "Pedro Alves Cabral", "(94) 99999-2222", True]
        }
        arq_vagas = open("vagas.dat", "wb")
        pickle.dump(vagas, arq_vagas)
        arq_vagas.close()
    return vagas




def grava_clientes(clientes):
    arq_clientes = open("clientes.dat", "wb")
    pickle.dump(clientes, arq_clientes)
    arq_clientes.close()

def grava_locacoes(locacoes):
    arq_locacoes = open("locacoes.dat", "wb")
    pickle.dump(locacoes, arq_locacoes)
    arq_locacoes.close()

def grava_vagas(vagas):
    arq_vagas = open("vagas.dat", "wb")
    pickle.dump(vagas, arq_vagas)
    arq_vagas.close()


clientes = recuperar_clientes()
locacoes = recuperar_locacoes()
vagas = recuperar_vagas()

op_princ = ""
while op_princ != "0":

  print("################################################")
  print("##### Sistema de Gestão de Estacionamentos #####")
  print("################################################")
  print("########       1 - Módulo Clientes      ########")
  print("########       2 - Módulo Locação       ########")
  print("########       3 - Módulo Vagas         ########")
  print("########       4 - Relatório            ########")
  print("########       5 - Sobre o Sistema      ########")
  print("########       0 - Sair                 ########")
  op_princ = input("######## Escolha uma opção acima: ")
  print()
  if op_princ == "1":
    op_cliente = ""
    while op_cliente != "0":
       print("#########################################################")
       print("############ Você está no Módulo de Clientes ############")
       print("########       1 - Cadastrar Cliente             ########")
       print("########       2 - Exibir dados do Cliente       ########")
       print("########       3 - Alterar dados do Cliente      ########")
       print("########       4 - Remover Cliente               ########")
       print("########       0 - Retornar ao Menu Principal    ########")
       print()
       op_cliente = input("######## Escolha sua opção: ")
       print()
       if op_cliente == "1":
         print("#######################################################")
         print("################## Cadastrar Cliente ##################")
         print("#######################################################")
         print()
         nome = input("Informe seu nome: ")
         print()
         email = input("Informe seu e-mail: ")
         print()
         data_nasc = input("Informe sua data de nascimento: ")
         print()
         cel = input("Informe seu número de celular: ")
         print()
         clientes[nome] = [email, data_nasc, cel, True]
         print("Clientes: ", clientes)
         print("Cliente cadastrado com sucesso! ")
         print()
         print("Tecle  para continuar...")
         print()

       elif op_cliente == "2":
           print("#############################################################")
           print("################## Exibir Dados do Cliente ##################")
           print("#############################################################")
           print()
           nome = input("######## Insira o seu nome: ")
           print()
           if nome in clientes:
             print()
             print("######## E-mail: ", clientes[nome][0])
             print("######## Data de nascimento: ", clientes[nome][1])
             print("######## Celular: ", clientes[nome][2])
             print("######## Cliente ativo/desativado: ", clientes[nome][3])
           else:
             print("Cliente não encontrado!")
           print()
           print("Tecle  para continuar...")
           print()

       elif op_cliente == "3":
             print("#############################################################")
             print("################## Alterar Dados do Cliente #################")
             print("#############################################################")
             print()
             nome = input("######## Nome: ")
             if nome in clientes:
               print("######## Dados atuais do cliente: ")
               print("######## E-mail: ", clientes[nome][0])
               print("######## Data de nascimento: ", clientes[nome][1])
               print("######## Celular: ", clientes[nome][2])
               print()
               print("######## Digite os novos dados para o cliente: ")
               email = input("######## Novo e-mail: ")
               data_nasc = input("######## Nova data de nascimento: ")
               cel = input("######## Novo número de celular: ")
               status = clientes[nome][3]
               clientes[nome] = [email, data_nasc, cel, status]
               print()
               print("Dados do cliente alterados com sucesso!")
               print()
               print("Clientes: ", clientes)
               print()
             else:
               print("Não existe nenhum cliente com esse nome!")
             print()
             input("Tecle  para continuar...")
             print()

       elif op_cliente == "4":
            print("#############################################################")
            print("################## Remover Dados do Cliente #################")
            print("#############################################################")
            print()
            nome = input("######## Nome: ")
            if nome in clientes:
               print("######## Cliente encontrado:")
               print("######## E-mail: ", clientes[nome][0])
               print("######## Data de nascimento: ", clientes[nome][1])
               print("######## Celular: ", clientes[nome][2])
               print()
               confirma = input("######## Deseja desativar esse cliente? S/N ")
               if confirma == "S":
                 clientes[nome][3] = False
                 print("Cliente desativado com sucesso!")
                 print()
                 print("Clientes: ", clientes)
               else:
                 print("Cliente não foi desativado!")
            else:
             print("Não existe nenhum cliente com esse nome!")
            print()
            input("Tecle  para continuar...")
            print()

       elif op_cliente == "0":
         print("###################################################")
         print("###################################################")
         print("###################################################")
         print("#######                                     #######")
         print("#####             Voltando ao                 #####")
         print("#####            Menu Principal...            #####")
         print("#######                                     #######")
         print("###################################################")

    print()
    input("Tecle  para continuar...")
    print()




  elif op_princ == "2":
    op_locacao = ""
    while op_locacao != "0":
        print("#########################################################")
        print("############ Você está no Módulo de Locação  ############")
        print("########       1 - Fazer Locação                 ########")
        print("########       2 - Pesquisar Locação             ########")
        print("########       3 - Alterar dados da Locação      ########")
        print("########       4 - Cancelar Locação              ########")
        print("########       0 - Retornar ao Menu Principal    ########")
        print()
        op_locacao = input("######## Escolha sua opção: ")
        print()
        if op_locacao == "1":
          print("#######################################################")
          print("##################  Fazer a Locação  ##################")
          print("#######################################################")
          print()
          cpf = input("########Informe seu CPF: ")
          print()
          nome = input("########Informe seu Nome Completo: ")
          print()
          email = input("########Informe seu e-mail: ")
          print()
          data_nasc = input("########Informe sua data de nascimento: ")
          print()
          cel = input("########Informe seu número de celular: ")
          print()
          locacoes[cpf] = [nome, email, data_nasc, cel, True]
          print("Locações: ", locacoes)
          print("Locação feita com sucesso!")
          print()
          print("Pressione  para continuar...")
          print()


        elif op_locacao == "2":
          print("#############################################################")
          print("################## Exibir Dados da Locação ##################")
          print("#############################################################")
          print()
          cpf = input("######## Insira o seu CPF: ")
          print()
          if cpf in locacoes:
            print("######## Nome: ", locacoes[cpf][0])
            print("######## E-mail: ", locacoes[cpf][1])
            print("######## Data de nascimento: ", locacoes[cpf][2])
            print("######## Celular: ", locacoes[cpf][3])
            print("######## Locação ativa/desativa: ", locacoes[cpf][4])
          else:
            print("Não existe nenhuma locação com esse CPF!")
          print()
          input("Tecle  para continuar...")
          print()

        elif op_locacao == "3":
          print("#############################################################")
          print("################## Alterar Dados da Locação #################")
          print("#############################################################")
          print()
          cpf = input("######## CPF: ")
          print()
          if cpf in locacoes:
            print("######## Dados atuais da locação: ")
            print("######## Nome completo: ", locacoes[cpf][0])
            print("######## E-mail: ", locacoes[cpf][1])
            print("######## Data de nascimento: ", locacoes[cpf][2])
            print("######## Celular: ", locacoes[cpf][3])
            print()
            print("######## Digite os novos dados para a locação: ")
            nome = input("######## Novo nome: ")
            email = input("######## Novo e-mail: ")
            data_nasc = input("######## Nova data de nascimento: ")
            cel = input("######## Novo número de celular: ")
            status = locacoes[cpf][4]
            locacoes[cpf] = [nome, email, data_nasc, cel, status]
            print()
            print("Dados da locação alterados com sucesso!")
            print()
            print("Locações: ", locacoes)
            print()
          else:
            print("Não existe nenhuma locação com esse CPF!")
          print()
          input("Tecle  para continuar...")
          print()

        elif op_locacao == "4":
            print("#############################################################")
            print("##################    Cancelar a Locação    #################")
            print("#############################################################")
            print()
            cpf = input("######## CPF: ")
            if cpf in locacoes:
              print("######## Locação encontrada:")
              print("######## Nome completo: ", locacoes[cpf][0])
              print("######## E-mail: ", locacoes[cpf][1])
              print("######## Data de nascimento: ", locacoes[cpf][2])
              print("######## Celular: ", locacoes[cpf][3])
              print()
              confirma = input("######## Deseja cancelar essa locação? S/N ")
              if confirma == "S":
                locacoes[cpf][4] = False
                print("Locação cancelada com sucesso!")
                print()
                print("Locações: ", locacoes)
              else:
                print("Cancelamento não foi concluído!")
            else:
              print("Não existe nenhuma locação com esse CPF!")
            print()
            input("Tecle  para continuar...")
            print()

        elif op_locacao == "0":
          print("###################################################")
          print("###################################################")
          print("###################################################")
          print("#######                                     #######")
          print("#####             Voltando ao                 #####")
          print("#####            Menu Principal...            #####")
          print("#######                                     #######")
          print("###################################################")

    print()
    input("Tecle  para continuar...")
    print()


  elif op_princ == "3":
    op_vagas = ""
    while op_vagas != "0":
        print("####################################################")
        print("#########   Você está no Módulo de Vagas   #########")
        print("####################################################")
        print("########       1 - Cadastrar Vaga                ########")
        print("########       2 - Pesquisar Vaga                ########")
        print("########       3 - Alterar dados da Vaga         ########")
        print("########       4 - Desocupar Vaga                ########")
        print("########       0 - Retornar ao Menu Principal    ########")
        print()
        op_vagas = input("######## Escolha sua opção: ")
        if op_vagas == "1":
          print("#######################################################")
          print("##################   Cadastrar Vaga  ##################")
          print("#######################################################")
          print()
          placa = input("########Informe a placa do seu carro: ")
          print()
          carro = input("########Informe o modelo do seu carro: ")
          print()
          cor = input("########Informe a cor do seu carro: ")
          print()
          nome = input("########Informe o seu nome completo: ")
          print()
          cel = input("########Informe seu número de celular: ")
          print()
          vagas[placa] = [carro, cor, nome, cel, True]
          print("Vagas: ", vagas)
          print("Vaga cadastrada com sucesso!")
          print()
          print("Pressione  para continuar...")
          print()


        elif op_vagas == "2":
          print("#############################################################")
          print("################## Exibir Dados da Vaga    ##################")
          print("#############################################################")
          print()
          placa = input("######## Insira a placa do seu carro: ")
          print()
          if placa in vagas:
            print("######## Carro: ", vagas[placa][0])
            print("######## Cor: ", vagas[placa][1])
            print("######## Nome completo: ", vagas[placa][2])
            print("######## Celular: ", vagas[placa][3])
            print("######## Vaga ativa/desativa: ", vagas[placa][4])
          else:
            print("Não existe nenhuma vaga com essa placa cadastrada!")
          print()
          input("Tecle  para continuar...")
          print()

        elif op_vagas == "3":
          print("#############################################################")
          print("##################  Alterar Dados da Vaga   #################")
          print("#############################################################")
          print()
          placa = input("######## Placa: ")
          print()
          if placa in vagas:
            print("######## Dados atuais da vaga cadastrada: ")
            print("######## Carro: ", vagas[placa][0])
            print("######## Cor: ", vagas[placa][1])
            print("######## Nome completo: ", vagas[placa][2])
            print("######## Celular: ", vagas[placa][3])
            print()
            print("######## Digite os novos dados para a vaga: ")
            carro = input("######## Novo carro: ")
            cor = input("######## Nova cor: ")
            nome = input("######## Novo nome: ")
            cel = input("######## Novo número de celular: ")
            status = vagas[placa][4]
            vagas[placa] = [carro, cor, nome, cel, status]
            print()
            print("Dados da vaga alterados com sucesso!")
            print()
            print("Vagas: ", vagas)
            print()
          else:
            print("Não existe nenhuma vaga com essa placa cadastrada!")
          print()
          input("Tecle  para continuar...")
          print()

        elif op_vagas == "4":
            print("#############################################################")
            print("##################     Desocupar a Vaga     #################")
            print("#############################################################")
            print()
            placa = input("######## Placa: ")
            if placa in vagas:
              print("######## Vaga encontrada:")
              print("######## Carro: ", vagas[placa][0])
              print("######## Cor: ", vagas[placa][1])
              print("######## Nome completo: ", vagas[placa][2])
              print("######## Celular: ", vagas[placa][3])
              print()
              confirma = input("######## Deseja desocupar essa vaga? S/N ")
              if confirma == "S":
                vagas[placa][4] = False
                print("Vaga desocupada com sucesso!")
                print()
                print("Vagas: ", vagas)
              else:
                print("Desocupamento não foi concluído!")
            else:
              print("Não existe nenhuma vaga com essa placa cadastrada!")
            print()
            input("Tecle  para continuar...")
            print()

        elif op_vagas == "0":
          print("###################################################")
          print("###################################################")
          print("###################################################")
          print("#######                                     #######")
          print("#####             Voltando ao                 #####")
          print("#####            Menu Principal...            #####")
          print("#######                                     #######")
          print("###################################################")

    print()
    input("Tecle  para continuar...")
    print()

  elif op_princ == "4":
    op_relat = ""
    while op_relat != "0":
        print("################################################################")
        print("############### Você está no Módulo de Relatório ###############")
        print("########       1 - Listagem Geral de Dados              ########")
        print("########       2 - Lista de Locações Feitas             ########")
        print("########       3 - Lista de Vagas Ocupadas              ########")
        print("########       0 - Retornar ao Menu Principal           ########")
        print()
        op_relat = input("######## Escolha sua opção: ")
        print()
        if op_relat == "1":
          print("################################################################")
          print("################# Listagem Geral de Clientes ###################")
          print("################################################################")
          print()
          print(f"{'Nome':<20} | {'Email':<30} | {'Data Nasc.':<15} | {'Celular':<20} | {'Status':<5}")
          print("-" * 95)
          for nome, dados in clientes.items():
            email, data_nasc, cel, status = dados
            print(f"{nome:<20} | {email:<30} | {data_nasc:<15} | {cel:<20} | {str(status):<5}")
          print()

          print("################################################################")
          print("################## Listagem Geral de Locações ##################")
          print("################################################################")
          print()
          print(f"{'CPF':<15} | {'Nome':<20} | {'Email':<30} | {'Data Nasc.':<15} | {'Celular':<20} | {'Status':<5}")
          print("-" * 110)
          for cpf, dados in locacoes.items():
            nome, email, data_nasc, cel, status = dados
            print(f"{cpf:<15} | {nome:<20} | {email:<30} | {data_nasc:<15} | {cel:<20} | {str(status):<5}")
          print()

          print("################################################################")
          print("################### Listagem Geral de Vagas ####################")
          print("################################################################")
          print()
          print(f"{'Placa':<10} | {'Carro':<15} | {'Cor':<10} | {'Nome Cliente':<25} | {'Celular':<20} | {'Status':<7}")
          print("-" * 100)
          for placa, dados in vagas.items():
            carro, cor, nome_cliente, cel, status = dados
            print(f"{placa:<10} | {carro:<15} | {cor:<10} | {nome_cliente:<25} | {cel:<20} | {str(status):<7}")
          print()


    print()
    print("Retornando ao Menu Principal...")
    input("Tecle  para continuar...")
    print()

  elif op_princ == "5":
    print("######################################################")
    print("########  Você está no Módulo de Informações  ########")


    print("######################################################")
    print("#######  #######")
    print("#######    <-Curso: Sistemas de Informação->   #######")
    print("<---------------------------------------------------->")
    print("#####    ***Aluno que está desenvolvendo:      #######")
    print("#####         * Lucas Araújo Diniz  @UFRN      #######")
    print("<---------------------------------------------------->")
    print("#######  ***Sob orientação do professor:       #######")
    print("#######       *   Flavius Gorgônio             #######")
    print("######################################################")
    print()
    print("Retornando ao Menu Principal...")
    input("Tecle  para continuar...")
    print()

  elif op_princ == "0":
    print("###################################################")
    print("###################################################")
    print("###################################################")
    print("#######                                     #######")
    print("#####            Você encerrou o              #####")
    print("#####           programa, até logo!           #####")
    print("#######                                     #######")
    print("###################################################")

  else:
    print()
    print("###################################################")
    print("########   Você digitou uma opção inválida  #######")
    print("###################################################")
    print("########                                    #######")
    print("#######       Retorne ao menu anterior       ######")
    print("#######          e tente novamente           ######")
    print("########                                    #######")
    print("###################################################")
    print()
    print("Retornando ao Menu Principal...")
    input("Tecle  para continuar...")
    print()

print("Obrigado por usar o nosso programa!")


grava_clientes(clientes)
grava_locacoes(locacoes)
grava_vagas(vagas)
```
